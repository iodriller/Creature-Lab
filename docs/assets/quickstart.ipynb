## 5. Render the run report inline

`--html` writes a self-contained run card: score breakdown, signal sparklines, and
a root-path plot. Display it directly in the notebook.

```python
!creature-lab report latest --html report.html

from IPython.display import HTML, display

display(HTML(open("report.html").read()))
```

## 6. Improve it

`evolve` searches for a better body/gait locally (no cloud, no GPU) and saves the
best candidate plus its lineage. A handful of attempts is enough to see the score
move; the packaged zoo examples use `--attempts 20` or more for a real search.

```python
# `evolve` takes creature/task JSON file paths, so write the zoo's packaged
# quadruped + crawl_forward task out to local files first.
from creature_lab.zoo import zoo_creature

creature, task = zoo_creature("quadruped")
open("quadruped.json", "w").write(creature.model_dump_json(indent=2))
open("crawl_forward.json", "w").write(task.model_dump_json(indent=2))
```

```python
!creature-lab evolve quadruped.json --task crawl_forward.json --attempts 10
```
